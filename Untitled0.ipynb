```python
# FIND-S Algorithm
# Machine Learning Experiment

# Training dataset
# Attributes: Sky, AirTemp, Humidity, Wind, Water, Forecast
# Last column: EnjoySport (Yes/No)

training_data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Cool', 'Change', 'Yes']
]

# FIND-S algorithm
def find_s(data):
    # Initialize hypothesis with the most specific values
    hypothesis = ['0'] * (len(data[0]) - 1)

    print("Initial Hypothesis:", hypothesis)
    print()

    # Examine each training example
    for example in data:

        # Consider only positive examples
        if example[-1] == 'Yes':

            print("Positive Example:", example)

            # Compare each attribute
            for i in range(len(hypothesis)):

                # First positive example
                if hypothesis[i] == '0':
                    hypothesis[i] = example[i]

                # If values are different, generalize to '?'
                elif hypothesis[i] != example[i]:
                    hypothesis[i] = '?'

            print("Updated Hypothesis:", hypothesis)
            print()

    return hypothesis


# Run FIND-S
final_hypothesis = find_s(training_data)

# Display final result
print("Final Most Specific Hypothesis:")
print(final_hypothesis)
```
