```python
import csv

def candidate_elimination(csv_path):
    with open(csv_path) as f:
        data = list(csv.reader(f))
    concepts, target = [r[:-1] for r in data[1:]], [r[-1] for r in data[1:]]
    S, G = concepts[0].copy(), [['?'] * len(concepts[0]) for _ in range(len(concepts[0]))]

    for i, ex in enumerate(concepts):
        if target[i].lower() in ['yes', '1', 'positive', 'enjoy']:
            for j in range(len(S)):
                if ex[j] != S[j]: S[j] = G[j][j] = '?'
        else:
            for j in range(len(S)):
                G[j][j] = S[j] if ex[j] != S[j] else '?'

    return S, [g for g in G if g != ['?'] * len(S)]

# Demo
with open("data.csv", "w") as f:
    f.write("Sky,Temp,Hum,Wind,Water,Fcst,Enjoy\nSunny,Warm,Norm,Strong,Warm,Same,Yes\nSunny,Warm,High,Strong,Warm,Same,Yes\nRainy,Cold,High,Strong,Warm,Change,No\nSunny,Warm,High,Strong,Cool,Change,Yes")

S, G = candidate_elimination("data.csv")
print("S:", S)
print("G:", G)
```
