```python
import math, pandas as pd

def entropy(df, target):
    return sum(-p * math.log2(p) for p in df[target].value_counts(normalize=True))

def build_id3(df, target, feats):
    if len(df[target].unique()) == 1: return df[target].iloc[0]
    if not feats: return df[target].mode()[0]

    best = max(feats, key=lambda f: entropy(df, target) - sum(len(sub)/len(df) * entropy(sub, target) for _, sub in df.groupby(f)))
    return {best: {v: build_id3(sub, target, [f for f in feats if f != best]) for v, sub in df.groupby(best)}}

def predict(tree, sample):
    if not isinstance(tree, dict): return tree
    k = list(tree.keys())[0]
    return predict(tree[k].get(sample.get(k)), sample)

# Demo
df = pd.DataFrame({
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain', 'Overcast', 'Sunny'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Strong', 'Strong', 'Weak'],
    'Play': ['No', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No']
})

tree = build_id3(df, 'Play', list(df.columns[:-1]))
print("Tree:", tree)
print("Prediction:", predict(tree, {'Outlook': 'Rain', 'Wind': 'Strong'}))
```
